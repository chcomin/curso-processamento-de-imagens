## Numpy

```python
import numpy as np  # Importa biblioteca numpy, dando a ela o nome np

vector1 = [2.3, 5.4, 3.3, 2.5, 7.3]
vector2 = [1.2, 5.4, 2.3, 7.6, 4.3]

# Transforma lista em um "vetor numpy". Ao contrário de listas, 
# um vetor numpy possui dados contíguos na memória. Fundamentalmente, 
# é um array C que pode ser acessado pelo Python.
vector1 = np.array(vector1)
vector2 = np.array(vector2)

# Multiplica cada elemento do vetor v1 pelo número 5
result = 5*vector1

# Soma de vetores
vector_sum = vector1 + vector2

# Media e desvio padrao
mean = np.mean(vector1)
std = np.std(vector1)
```

```python
# Define matrizes (lista de listas)
matrix1 = [[3.2, 5.3, 6.2, 1.0],
           [7.4, 2.3, 5.3, 0.4],
           [4.3, 2.4, 7.5, 2.3],
           [0.5, 1.3, 4.5, 2.4]]

matrix2 = [[4.6, 2.5, 3.5, 2.6],
           [1.4, 5.3, 5.3, 3.4],
           [7.3, 2.5, 6.5, 9.3],
           [5.5, 2.5, 3.5, 8.4]]

# Produto de matrizes
matrix_product = np.dot(matrix1, matrix2)

# Soma elemento a elemento de matrizes
matrix_sum = matrix1 + matrix2
```

```python
# Resolve o seguinte sistema de equacoes:
# 3x + 5y  + 3z = 54.6
# x  + 12y + 5z = 98.8
# 5x + 7y  + 2z = 65.2

coeffs = [[3,  5, 3],
          [1, 12, 5],
          [5,  7, 2]]

right_values = [54.6, 98.8, 65.2]

result = np.linalg.solve(coeffs, right_values)
print(f"x={result[0]:.2}, y={result[1]:.2}, z={result[2]:.2}")

#Verifica solucao
prod = np.dot(coeffs, result)
print(f"Produto coefs solucao: {prod.T}")
```

## Matplotlib

```python
# Importa o submódulo pyplot (que possui as funções de plotar) com nome plt
import matplotlib.pyplot as plt

# Gera um vetor de valores entre 0 e 2*pi, com passo 0.01
x = np.arange(0, 2*np.pi, 0.01)   # x<-[0, 0.01, 0.02, ...]
# Aplica a função seno a cada elemento do vetor x, gerando um novo vetor y
y = np.sin(2*x)
plt.plot(x, y)
```
