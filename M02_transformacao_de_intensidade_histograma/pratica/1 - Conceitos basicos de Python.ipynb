# Elementos básicos da linguagem Python

```python
# Variável
value = 5
name = "Paulo"
flag = True 

print(value)
print(name)
print(flag)
```

```python
# Impressão de variáveis na tela
val1 = 10
val2 = 300
print(f"Os valores são: {val1} e {val2}")
```

```python
# Condicional
value = 10
if value<5:
    print("Valor menor do que 5")
else:
    print("Valor maior ou igual a 5")
    
# Múltiplas condições
value1 = 3
value2 = 10
if value1<5 and value2<5:
    print("Ambos valores menores que 5")
elif value1<5 and value2>=5:
    print("valor1 menor que 5 e valor2 maior ou igual a 5")
elif value1>=5 and value2<5:    
    print("valor1 maior ou igual a 5 e valor2 menor que 5")
else:
    print("valor1 e valor2 maiores que 5 ")
```

```python
# Lista de valores
my_list = [1, 2, 3, 4, 5, 6]
print(f"valores: {my_list}")

# Uma lista pode conter variáveis de tipos diferentes
my_list = [3, 4.5, True, "casa"]

# Lista de listas (matriz)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(f"matriz: {matrix}")

# Lista de valores inteiros entre 3 e 10
interval_list = list(range(3, 11))
print(f"lista_intervalo: {interval_list}")

## Manipulação de listas
my_list = [1, 2, 3, 4, 5, 6]

# Adiciona valor ao final da lista
my_list.append(6)

# Remove valor no final
last_value = my_list.pop()

# Remove valor no indice 3
value = my_list.pop(3)

# Remove primeira ocorrência do número 2
my_list.remove(2)
print(f"minha_lista: {my_list}")

# Fatiamento de lista
print(my_list[2:4])
```

```python
# Loop for
print("Resultado do laço for:")
values = [4, 8, 2, 6, -5, 10]
for index in range(len(values)):
    print(values[index])
    
print("---")  

# Loop for "pythonico"
print('Resultado do loop for "pythonico":')
for val in values:
    print(val)

print("---")   
# Loop while
print("Resultado do loop while:")
value = 0
while value<10:
    value += 1
    print(value)
```

```python
# Função
def add(v1, v2):
    result = v1 + v2
    return result

print(add(3, 4))
```
