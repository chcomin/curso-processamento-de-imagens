# Leitura e plotagem de imagens utilizando Python

```python
# Biblioteca padrão do Python para manipulação de arquivos e diretórios.
from pathlib import Path

# Biblioteca matplotlib, utilizada para plotar figuras. 
import matplotlib.pyplot as plt

# Biblioteca numpy utilizada para manipulação de matrizes e operações matemáticas.
import numpy as np

# Biblioteca PIL (Python Imaging Library) utilizada para manipulação de imagens.
from PIL import Image

# Diretório onde estão os dados utilizados nos notebooks.
image_file = Path("../../dados/earth.jpg")
```

```python
# Leitura da imagem utilizando o matplotlib. A imagem retornada é um array da biblioteca numpy
img = plt.imread(image_file)
# Plotagem da imagem
plt.imshow(img)
```

É sempre interessante verificarmos algumas informações sobre a imagem lida:

```python
print("Objeto:", type(img))
print("Tipo de dado de cada pixel:", img.dtype)
print("Tamanho da imagem:", img.shape)
print("Valor máximo:", img.max())
print("Valor mínimo:", img.min())
```

No caso, a imagem é:

* Possui 800 linhas e 1200 colunas
* Possui 3 canais

```python
# Leitura da imagem utilizando a biblioteca pillow
img = np.array(Image.open(image_file))
```

```python
# Podemos imprimir o valor de um pixel específico da imagem:
print(img[400, 560])
```

# Extração e visualização de um canal de cor

```python
img_r = img[:,:,0]   # Sintaxe do numpy para extração do canal 0 (vermelho)
plt.imshow(img_r)
```

Porque obtemos essa imagem azul e verde?

Ao visualizarmos uma matriz de valores no matplotlib, a biblioteca não sabe que queremos visualizar uma imagem. Por padrão, a biblioteca mapeia os valores da matriz usando um **mapa de cores**. Esse mapa de cores gera cores falsas para representar os valores da matriz. Para o que queremos fazer, é mais interessante visualizar os valores em níveis de cinza. Para isso, fazemos o seguinte:

```python
plt.imshow(img_r, cmap="gray")
```

Vemos que regiões da imagem original nas quais o vermelho é mais predominante (continentes) se tornam brancas nessa imagem em nível de cinza do canal vermelho.
