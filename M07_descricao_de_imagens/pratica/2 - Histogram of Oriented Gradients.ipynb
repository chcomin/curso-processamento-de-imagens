```python
import matplotlib.pyplot as plt
from skimage.feature import hog

img = plt.imread("../../dados/cameraman.tiff")

features, hog_image = hog(
    img,
    # Número de caixas do histograma de orientações de graidente
    orientations=9,
    # Tamanho da célula em pixels
    pixels_per_cell=(16, 16),
    # Número de células por bloco
    cells_per_block=(2, 2),
    visualize=True
)

plt.figure(figsize=(10,10))
plt.imshow(hog_image, "gray", vmax=10)
```
