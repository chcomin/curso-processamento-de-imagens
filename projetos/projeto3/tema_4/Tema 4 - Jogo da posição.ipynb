# Jogo da posição

- O programa mostra um esqueleto na tela sobre a imagem da webcam. A pessoa possui X segundos para fazer a mesma posição
- Se o tempo acabar ou a pessoa conseguir fazer a posição, uma nova posição é mostrada
- Usar o modelo YOLO de detecção de pose
- Tratar possíveis erros e situações inesperadas (ex: outra pessoa aparecer na câmera)

Para gerar posições alvo de esqueleto, abrir o arquivo [skeleton.html](skeleton.html) em um navegador (testado no Chrome) para extrair coordenadas de pontos chave.

Exemplo de uso do código:

```python
from stream import VideoStream
from engine import GameEngine

game_engine = GameEngine(
    distance_threshold=10,
    position=(300, 200),
    height=300
    )
video_stream = VideoStream(cam_id=0)
video_stream.start(game_engine)
```
